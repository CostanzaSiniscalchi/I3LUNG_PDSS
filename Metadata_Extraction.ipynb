# a) Metadata plots - single center

This script visualizes the distribution of a selected metadata column
from the extracted DICOM metadata Excel file. It can be used for any
categorical metadata field (e.g., Manufacturer, Slice Thickness, Contrast information, etc..). 
Metadata plots are created for each single center, separately.

## 1. INT

## Manufacturer

```python
import matplotlib.pyplot as plt
import pandas as pd
import os

def plot_metadata_distribution(file_path, column_name, ylim=None):
    """
    Args:
        file_path (str): Path to the Excel file containing metadata.
        column_name (str): Column to analyze (e.g. 'Manufacturer').
        ylim (tuple or None): Optional y-axis limits (min, max).          
    """

    # Load metadata
    df = pd.read_excel(file_path)
    df.columns = df.columns.str.strip()

    if column_name not in df.columns:
        raise ValueError(f"Column '{column_name}' not found in the dataset.")

    # Replace missing values (NaN or empty strings) with 'Missing Information'
    df[column_name] = df[column_name].replace('', pd.NA)
    df[column_name] = df[column_name].fillna('Missing Information')

    # Count unique values (including 'Missing Information')
    value_counts = df[column_name].value_counts()
    total = len(df)

    # Plot
    plt.figure(figsize=(10, 6), dpi=600)
    bars = value_counts.plot(kind='bar', color='skyblue')

    # Add value labels on bars
    for i, count in enumerate(value_counts):
        plt.text(i, count + max(value_counts) * 0.02, str(count),
                 ha='center', va='bottom', fontsize=12)

    plt.title(f"{column_name}", fontsize=16)
    plt.xlabel('')
    plt.ylabel('Number of CTs', fontsize=13)
    plt.xticks(rotation=30, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    if ylim:
        plt.ylim(ylim)

    plt.tight_layout()
    plt.show()

# === Example usage ===
if __name__ == "__main__":
    excel_path = "data/INT_metadata.xlsx"
    plot_metadata_distribution(
        file_path=excel_path,
        column_name='Manufacturer',
        ylim=(0, 250)
    )
```

# b) Metadata plots - all centers together

This script merges metadata from multiple centers and plots the distribution of a selected metadata field (for example, Manufacturer). You can change this field to any metadata variable you want to visualize, such as Contrast/Bolus Agent.

# Manufacturer

```python
import matplotlib.pyplot as plt
import pandas as pd

def plot_combined_metadata(file_paths, column_name, ylim=None):
    """
    Args:
        file_paths (list[str]): List of Excel files, one per center.
        column_name (str): Column name to analyze (e.g. 'Manufacturer').
        ylim (tuple or None): Optional y-axis limits.
    """

    # === LOAD AND MERGE DATA ===
    all_dfs = []
    for path in file_paths:
        df_temp = pd.read_excel(path)
        df_temp.columns = df_temp.columns.str.strip()
        all_dfs.append(df_temp)

    df = pd.concat(all_dfs, ignore_index=True)

    # === CHECK & CLEAN ===
    if column_name not in df.columns:
        raise ValueError(f"Column '{column_name}' not found in the merged dataset.")

    df[column_name] = df[column_name].astype(str).str.strip()
    df[column_name] = df[column_name].replace(['nan', 'None', 'NaN', ''], 'Missing information')

    # === COUNT UNIQUE VALUES ===
    value_counts = df[column_name].value_counts(dropna=False)
    total_cts = len(df)
    print(f"Total CT scans across all centers: {total_cts}")

    # === PLOT ===
    plt.figure(figsize=(10, 6), dpi=600)
    bars = value_counts.plot(kind='bar', color='skyblue')

    # Add value labels above bars
    for i, count in enumerate(value_counts):
        plt.text(i, count + max(value_counts) * 0.01, str(count),
                 ha='center', va='bottom', fontsize=12)

    plt.title(f"{column_name} (All Centers)", fontsize=16)
    plt.xlabel('')
    plt.ylabel('Number of CT Scans', fontsize=13)
    plt.xticks(rotation=30, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    if ylim:
        plt.ylim(ylim)

    plt.tight_layout()
    plt.show()


# === EXAMPLE USAGE ===
if __name__ == "__main__":
    file_paths = [
        "data/INT_metadata.xlsx",
        "data/GHD_metadata.xlsx",
        "data/VHIO_metadata.xlsx",
        "data/UOC_metadata.xlsx",
        "data/SZMC_metadata.xlsx",
        "data/MH_metadata.xlsx"
    ]

    plot_combined_metadata(
        file_paths=file_paths,
        column_name='Manufacturer',
        ylim=(0, 600)
    )
```
